```python
import cv2
```

```python
from google.colab.patches import cv2_imshow
img = cv2.imread("lion.jpg")
```

```python
cv2_imshow(img)
```

```python
print("SHAPE", img.shape)
```

```python
(b, g, r) = img[100, 50]
print("Blue", b, "green", g, "Red", r)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
cv2_imshow(gray)
cv2_imshow(gray2)
cv2_imshow(img)
```

```python
resize = cv2.resize(gray, (300, 200))
cv2_imshow(resize)
```

```python
cropped = img[100:300, 250:450]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(img, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("graylion.jpg", gray)
```

```python
img2 = cv2.imread("graylion.jpg")
cv2_imshow(img2)
```
